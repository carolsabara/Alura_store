```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

```

## Leitura da Base de dados




```python
urls = [
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv" ]

#list comprehension = uma forma compacta de criar uma lista em Python
lojas = [pd.read_csv(url) for url in urls]
loja1, loja2, loja3, loja4 = lojas
```

# Limpeza das três tabelas de uma única vez


```python
# Limpeza: verificar duplicados e nulos
for i, loja in enumerate(lojas, start=1):
    print(f"\n🔍 Loja {i}:")
    print(f"Duplicados: {loja.duplicated().sum()}")
    print(f"Nulos:\n{loja.isnull().sum()}")
```

```python
for i, loja in enumerate(lojas, start=1):
  print(f"\n Tipos de dados da Loja {i}")
  print(loja.dtypes)
```

```python
#alterar o tipo da data de object para datetime das tres tabelas
for loja in lojas :
  loja['Data da Compra'] = pd.to_datetime(loja['Data da Compra'], dayfirst=True, errors='coerce')

loja1.dtypes
```

```python
loja1.head(5)
```

#1.Faturamento de cada loja

```python
fat_loja1 = (loja1['Preço'].sum())
print(f"R$ {fat_loja1:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))
```

```python
fat_loja2 = (loja2['Preço'].sum())
print(f'R$ {fat_loja2:,.2f}'.replace(",", "X").replace(".", ",").replace("X", "."))
```

```python
fat_loja3 = (loja3['Preço'].sum())
print(f"R$ {fat_loja3:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))
```

```python
fat_loja4 = (loja4['Preço'].sum())
print(f'R$ {fat_loja4:,.2f}'.replace(",", "X").replace(".", ",").replace("X", "."))
```

```python
plt.figure(figsize=(8,5))

# Definir os rótulos das lojas e os valores de faturamento
labels = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']
faturamentos = [fat_loja1, fat_loja2, fat_loja3, fat_loja4]

plt.bar(labels, faturamentos, color='royalblue')
plt.title('Faturamento total por loja')
plt.ylabel('Faturamento (R$)')
plt.grid(axis='y', linestyle='--', alpha=0.6)

# Formatar os rótulos do eixo Y com pontos de milhar e vírgula decimal
formatter = mticker.FuncFormatter(lambda x, _: f'R$ {x:,.0f}'.replace(',', 'X').replace('.', ',').replace('X', '.'))
plt.gca().yaxis.set_major_formatter(formatter)


plt.tight_layout()
plt.show()
```

```python

# DataFrame de comparação
df_faturamento = pd.DataFrame({
    'Loja': ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4'],
    'Faturamento (R$)': [fat_loja1, fat_loja2, fat_loja3, fat_loja4]
})

# Formatar valores no padrão brasileiro
df_faturamento['Faturamento (R$)'] = df_faturamento['Faturamento (R$)'].map(lambda x: f"R$ {x:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))


display(df_faturamento)
```

# 2. Vendas por Categoria

```python
vendas_por_categoria3 = loja3.groupby('Categoria do Produto')['Preço'].sum().sort_values(ascending=False)
vendas_por_categoria3 = vendas_por_categoria3.apply(lambda x: f"R$ {x:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))
print(vendas_por_categoria3)
```

```python
vendas_por_categoria4 = loja4.groupby('Categoria do Produto')['Preço'].sum().sort_values(ascending=False)
vendas_por_categoria4 = vendas_por_categoria4.apply(lambda x: f"R$ {x:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))
print(vendas_por_categoria4)
```

```python
def formatar_brl(x):
    return f"R$ {x:,.2f}".replace(",", "X").replace(".", ",").replace("X", ".")

df_comparativo = (
    pd.DataFrame({
        f'Loja {i}': loja.groupby('Categoria do Produto')['Preço'].sum()
        for i, loja in enumerate([loja1, loja2, loja3, loja4], 1)
    })
    .loc[lambda df: df.sum(axis=1).sort_values(ascending=False).index]
    .apply(lambda col: col.map(formatar_brl))  # <- substitui applymap
)

display(df_comparativo)
```

# 5. Frete Médio por Loja

```python
# Calcular a média de frete em cada loja
frete_medio = [
    loja1['Frete'].mean(),
    loja2['Frete'].mean(),
    loja3['Frete'].mean(),
    loja4['Frete'].mean()
]

labels = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']

# Criar o gráfico de linha
plt.figure(figsize=(8,5))
plt.plot(labels, frete_medio, marker='o', color='mediumseagreen', linewidth=2)

# Títulos e rótulos
plt.title('Média de Frete por Loja')
plt.ylabel('Valor Médio do Frete (R$)')
plt.ylim(auto=True) # Ajustar o limite inferior do eixo y automaticamente
plt.grid(True, linestyle='--', alpha=0.6)

# Adicionar rótulos de valor sobre os pontos
for i, v in enumerate(frete_medio):
    plt.text(i, v + 0.05, f"{v:.2f}", ha='center', fontsize=9)

plt.tight_layout()
plt.show()
```

```python
resumo_final = pd.DataFrame({
    'Loja': ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4'],
    'Faturamento Total (R$)': [
        fat_loja1,
        fat_loja2,
        fat_loja3,
        fat_loja4
    ],
    'Média de Avaliação': [
        avaliação1,
        avaliação2,
        avaliação3,
        avaliação4
    ],
    'Frete Médio (R$)': [
        frete_medio1,
        frete_medio2,
        frete_medio3,
        frete_medio4
    ]
})

print(resumo_final)
```

```python
total_faturamento = resumo_final['Faturamento Total (R$)'].sum()

resumo_final['Participação no Faturamento (%)'] = (
    resumo_final['Faturamento Total (R$)'] / total_faturamento * 100
).round(2)

resumo_final = resumo_final.sort_values(by='Faturamento Total (R$)')
display(resumo_final)



```
